```python
import tensorflow as tf
import matplotlib.pyplot as plt
import matplotlib.image as mpimg # load image from dir for imshow
import numpy as np
import os, random
from tensorflow.keras import layers, models, callbacks
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.utils import load_img, img_to_array
from tensorflow.keras.applications import EfficientNetB0
from sklearn.metrics import confusion_matrix, classification_report, ConfusionMatrixDisplay
```

```python
# put data in train_ds and val_ds, then view
data_path = "train"
img_size = (224, 224) # if using ImageNet, img_size have to be 224,244
batch_size = 32
seed = 44

train_ds = tf.keras.utils.image_dataset_from_directory(
    data_path,
    labels='inferred',
    label_mode='int',
    validation_split=0.2,
    subset='training',
    batch_size=batch_size,
    image_size=img_size,
    shuffle=True,
    seed=seed
)

class_names = train_ds.class_names

val_ds = tf.keras.utils.image_dataset_from_directory(
    data_path,
    labels='inferred',
    label_mode='int',
    validation_split=0.2,
    subset='validation',
    image_size=img_size,
    batch_size=batch_size,
    shuffle=True,
    seed=seed
)

for cls in os.listdir(data_path):
    print(cls,':', len(os.listdir(os.path.join(data_path, cls))), 'images')
```

```python
# function random 1 image
def view_random_image(target_dir, target_class):
    # target_folder = os.listdir(data_path)
    target_folder = target_dir + '/' + target_class  + '/'
    random_img = random.sample(os.listdir(target_folder),1)
    img = mpimg.imread(target_folder + random_img[0])

    print(f'Image shape: {img.shape}')
    plt.imshow(img)
    plt.title(target_class)
    plt.axis('off')
    
    return random_img
```

```python
view_random_image(target_dir='train', target_class='steak')
```

```python
# check images in train_ds
for image, label in train_ds.take(1):
    plt.figure(figsize=(12,12))
    for i in range(16):
        ax = plt.subplot(4, 4, i+1)
        plt.imshow(image[i].numpy().astype('uint8'))
        plt.title(train_ds.class_names[label[i]])
        plt.axis('off')
```

```python
# Increase images
data_aug = tf.keras.Sequential([
    layers.RandomFlip('horizontal'),
    layers.RandomRotation(0.2),
    layers.RandomZoom(0.2),
    layers.RandomBrightness(0.2),
    layers.RandomContrast(0.2)
])

normalization_layer = layers.Rescaling(1./255)
```

```python
# Build CNN model, this is basic, not using this
model = models.Sequential([
    data_aug,
    normalization_layer,
    layers.Conv2D(32, (3,3), activation='relu', input_shape=(*img_size,3)),
    layers.MaxPooling2D(2,2),

    layers.Conv2D(64, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),
    
    layers.Conv2D(128, (3,3), activation='relu'),
    layers.MaxPooling2D(2,2),

    layers.Flatten(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.3),
    layers.Dense(len(class_names), activation='softmax')            
])
```

```python
# Build CNN model base on pretrainedmodel Imagenet (rec)
base_model = EfficientNetB0(
    include_top=False, # Remove the original classifier head from ImageNet (to take this dataset classes name)
    input_shape=(256,256,3), # Input shape for our dataset (RGB images)
    weights='imagenet' # Load pretrained weights from ImageNet (trained on ~14M images)
)
base_model.trainable = False #  Freeze convolutional base to reuse pretrained ImageNet features

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),  # Convert feature maps to a single 1D vector per image
    layers.Dropout(0.3), # Regularization to prevent overfitting
    layers.Dense(len(class_names), activation='softmax') # Custom classifier for our dataset (e.g., pizza vs steak)
])

# check download from ImageNet
model = EfficientNetB0(weights='imagenet')
print("Model loaded successfully ✅")

# if using ImageNet, img_size have to be 224,244
```

```python
# set model compilation
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

early_stop = EarlyStopping(
    monitor='val_loss',
    patience=3,
    restore_best_weights=True
)

best_epoch_save = ModelCheckpoint(
    'best_model.keras',
    monitor='val_loss',
    mode='min',
    save_best_only=True,
    verbose=1
)
```

```python
# run training model
history_model = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=30,
    callbacks=[best_epoch_save,early_stop],
    verbose=1
)
```

```python
#
acc = history_model.history['accuracy']
val_acc = history_model.history['val_accuracy']

loss = history_model.history['loss']
val_loss = history_model.history['val_loss']

best_model = tf.keras.models.load_model('best_model.keras')
best_model.summary()
```

```python
# test model
test_ds = tf.keras.utils.image_dataset_from_directory(
    "test",
    image_size=img_size,
    batch_size=32,
    shuffle=False # important
)
# if test_ds is shuffled, batched wrong, or points to train_ds or val_ds wrongly,
# then y_true and y_pred are no longer in the same order, leading to confusion matrix.

loss, acc = best_model.evaluate(test_ds)
print(f"Test accuracy: {acc*100:.2f}%")
```

```python
# check model, non labels forder
test_dir = 'real_test/'
img_paths = [os.path.join(test_dir, f) for f in os.listdir(test_dir) if f.lower().endswith(('.jpg', '.jpeg', '.png'))]

cols = 3
rows = (len(img_paths) + cols - 1) // cols  # chia tròn lên
plt.figure(figsize=(12, 4 * rows))

for i, img_path in enumerate(img_paths[:12]):
    img = load_img(img_path, target_size=img_size)
    img_array = img_to_array(img)
    img_array = np.expand_dims(img_array, 0)
    
    pred = best_model.predict(img_array, verbose=0)
    pred_class = np.argmax(pred, axis=1)[0]
    confidence = np.max(pred)

    plt.subplot(rows, cols, i+1)
    plt.imshow(load_img(img_path))
    plt.title(f'{class_names[pred_class]} {confidence*100:.2f}%')
    plt.axis('off')

#plt.show()
```

```python
# Lấy nhãn thật và dự đoán
y_true = np.concatenate([y for x, y in test_ds], axis=0) # Lấy toàn bộ nhãn thật (true label) từ dataset test
# test_ds là dataset dạng (batch_images, batch_labels)
# [y for x, y in test_ds] lấy toàn bộ các batch nhãn y
# np.concatenate(..., axis=0) ghép các batch lại thành 1 mảng 1D, ví dụ [0, 1, 1, 0, 1, ...]

y_pred = np.argmax(best_model.predict(test_ds), axis=1) # Lấy nhãn dự đoán (predicted label) của model
# best_model.predict(test_ds) cho ra xác suất dạng (500, 2) nếu có 2 class
# Ví dụ: [0.7, 0.3] nghĩa là 70% pizza, 30% steak
#np.argmax(..., axis=1) chọn class có xác suất cao nhất → [0, 1, 0, 1, ...].

# Báo cáo chi tiết
print(classification_report(y_true, y_pred, target_names=class_names))

# Vẽ confusion matrix
cm = confusion_matrix(y_true, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=class_names)
disp.plot(cmap='Blues', values_format='d')
plt.title("Confusion Matrix - Test Set")
plt.show()
```

```python
file_paths = test_ds.file_paths # take img path
images = np.concatenate([x for x, y in test_ds], axis=0)
wrong_pred = np.where(y_pred != y_true)[0] # [0] to convert to array
print(f'Wrong count: {len(wrong_pred)}')

pred = best_model.predict(test_ds)
cols = 3
rows = (len(wrong_pred) + cols - 1) // cols  # chia tròn lên
plt.figure(figsize=(12,12))
for i, idi in enumerate(wrong_pred):
    confidence = np.max(pred[idi]) # confident %
    filename = os.path.basename(file_paths[idi])
    plt.subplot(cols, rows, i+1)
    plt.imshow(images[idi].astype('uint8'))
    plt.title(f'Filename: {filename} \nTrue: {class_names[y_true[idi]]} \nPred: {class_names[y_pred[idi]]} {confidence*100:.2f}%')
    plt.axis('off')
```
